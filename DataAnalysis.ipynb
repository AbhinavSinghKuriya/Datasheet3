```python
# function for calculating which product occur multiple times and how much time it occur
def duplicates(column_name):
    counts = df[column_name].value_counts()
    duplicates = counts[counts > 1]
    result = duplicates.reset_index()
    result.columns = [column_name, 'Count']
    a = result
    print(a)
duplicates('Make')
duplicates('Model')
duplicates('Vehicle_Class')
duplicates('Transmission')
```

```python
#plotting each numerical column relationship to each other
grid = sns.pairplot(df,corner=True)
```

Now we will extract the usefull columns that have meanings from this

```python
#custom function for plotting specific plots from the grid 

def speciplot(y,x):
    single_ax = grid.axes[y][x]
    fig, ax = plt.subplots(figsize=(10,10))
    for collection in single_ax.collections:    
        offsets = collection.get_offsets()
        ax.scatter(
            offsets[:,0],
            offsets[:,1],
            s = 5
        )
    ax.set_xlabel(single_ax.get_xlabel())
    ax.set_ylabel(single_ax.get_ylabel())
    plt.show()

#some noticeably important plots
speciplot(4,1)
speciplot(5,1)
speciplot(6,1)
speciplot(5,2)
speciplot(6,2)
speciplot(4,3)
speciplot(5,3)
speciplot(6,3)
speciplot(6,5)
```

```python
#heatmap for only numerical data
numdf = df.select_dtypes(include='number')

corr = numdf.corr()

plt.figure(figsize=(10,8))

sns.heatmap(
    corr,
    annot=True,     
    cmap='coolwarm',
    fmt='.2f',       
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
```
